### Streaming and Batch

### Streaming

Most models can stream their output content while it is being generated. By displaying output progressively, streaming significatly imporves user experience, particulary for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunck in real-time.

```python
model.invoke("Write me a 200 words paragraph on Artificial Intelligence")
```

```python
model.stream("Write me a 200 words paragraph on Artificial Intelligence")
```

```python
### We don't neee to wait until the full text is being generated, we may display it
### Using | to show the content that is basically generated
for chunck in model.stream("Write me a 200 words paragraph on Artificial Intelligence"):
    print(chunck.text, end="|", flush=True)
```

```python
for chunck in model.stream("Why do parrots have colorful feathers?"):
    print(chunck.text, end="|", flush=True)
```

```python
### We have to wait until the text is generated
model.invoke("Why do parrots have colorful feathers?")
```

### Batch

Batching a collection of independent requests to a model can significally improve performance and reduce costs, as the processing cna be done in parallel:

```python
responses = model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
])

for response in responses:
    print(response)
```

```python
responses = model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"],
    config={
        'max_concurrency': 5, # Limit to 5 parallel calls
    })

for response in responses:
    print(response)
```
